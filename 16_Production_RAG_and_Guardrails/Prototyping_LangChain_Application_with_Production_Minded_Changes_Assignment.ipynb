#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.

##### ✅ Answer:


**Memory vs Disk trade-offs:**

- In-memory cache is fast but lost on restart
- Disk cache persists but slower
- Current demo uses memory - not production-ready

**Cache invalidation:**

- No expiration dates on cached data
- No way to update when PDF changes
- Could serve old/wrong info

**Concurrent access:**

- Memory cache doesn't handle many users well
- No shared cache for multiple servers
- Race conditions possible

**Cache size:**

- Grows forever, no cleanup
- No monitoring of what's actually useful

**Cold starts:**

- First users get slow responses
- No pre-loading common queries

````python
import os
import time
import statistics
from typing import Dict, List, Optional

from langchain_core.messages import HumanMessage, AIMessage
try:
    from langchain_core.messages import ToolMessage  # Optional in some LC versions
except Exception:
    ToolMessage = None

from langgraph_agent_lib import (
    create_langgraph_agent,
    ProductionRAGChain,
)

print("🚀 Activity #2: Advanced Agent Testing")
print("=" * 80)

# Detect existing simple agent and rag_chain from previous cells
simple_agent_available = 'simple_agent' in globals() and simple_agent is not None
rag_chain_available = 'rag_chain' in globals() and rag_chain is not None
print(f"Simple agent available: {simple_agent_available}")
print(f"RAG chain available: {rag_chain_available}")

# Try to locate a helpfulness agent constructor if present in the library
helpfulness_agent = None
try:
    # If a helper exists in the library, import and instantiate it here.
    # Placeholder import name; if not present, we continue without it.
    from langgraph_agent_lib.agents import create_helpfulness_langgraph_agent  # type: ignore
    if rag_chain_available:
        helpfulness_agent = create_helpfulness_langgraph_agent(
            model_name="gpt-4.1-mini",
            temperature=0.1,
            rag_chain=rag_chain,
        )
except Exception:
    helpfulness_agent = None

print(f"Helpfulness agent available: {helpfulness_agent is not None}")


def get_tool_calls(messages: List) -> List[str]:
    tool_names: List[str] = []
    for m in messages:
        # Collect tool calls from AI messages
        if isinstance(m, AIMessage):
            tool_calls = getattr(m, "tool_calls", None)
            if tool_calls:
                for tc in tool_calls:
                    # Different LC versions structure tool_calls slightly differently
                    name = (
                        tc.get("name")
                        if isinstance(tc, dict)
                        else getattr(tc, "name", None)
                    )
                    if name:
                        tool_names.append(name)
        # Optionally collect tool names from tool messages, if available
        if ToolMessage is not None and isinstance(m, ToolMessage):
            # ToolMessage has a name field in many versions
            name = getattr(m, "name", None)
            if name:
                tool_names.append(name)
    return tool_names


def run_agent(agent, query: str) -> Dict:
    start = time.time()
    resp = agent.invoke({"messages": [HumanMessage(content=query)]})
    elapsed = time.time() - start
    messages = resp.get("messages", [])
    final_text = messages[-1].content if messages else ""
    tools_used = get_tool_calls(messages)
    return {
        "elapsed": elapsed,
        "final_text": final_text,
        "messages": messages,
        "tools_used": tools_used,
    }


# Example: Test different query types
queries_to_test = [
    "What is the main purpose of the Direct Loan Program?",  # RAG-focused
    "What are the latest developments in AI safety?",  # Web search
    "Find recent papers about transformer architectures",  # Academic search
    "How do the concepts in this document relate to current AI research trends?"  # Multi-tool
]

results: List[Dict] = []

for query in queries_to_test:
    print(f"\n🔍 {query}")

    if simple_agent_available:
        try:
            r = run_agent(simple_agent, query)
            print(f"  SimpleAgent: {r['elapsed']:.2f}s | Tools: {r['tools_used']}\n  → {r['final_text'][:220]}…")
            r.update({"agent": "simple", "query": query})
            results.append(r)
        except Exception as e:
            print(f"  SimpleAgent error: {e}")

    if helpfulness_agent is not None:
        try:
            rh = run_agent(helpfulness_agent, query)
            print(f"  HelpfulnessAgent: {rh['elapsed']:.2f}s | Tools: {rh['tools_used']}\n  → {rh['final_text'][:220]}…")
            rh.update({"agent": "helpfulness", "query": query})
            results.append(rh)
        except Exception as e:
            print(f"  HelpfulnessAgent error: {e}")

# 2) Compare Agent Behaviors (basic summary)
if results:
    print("\n📊 Comparison Summary")
    print("-" * 80)
    by_key = {}
    for r in results:
        by_key.setdefault((r["query"], r["agent"]), []).append(r)
    for (query, agent), rs in by_key.items():
        avg_time = statistics.mean(x["elapsed"] for x in rs)
        tool_set = sorted({t for x in rs for t in x["tools_used"]})
        print(f"[{agent}] {query} → avg {avg_time:.2f}s | tools seen: {tool_set}")

# 3) Cache Performance Analysis
print("\n🗄️ Cache Performance Analysis")
print("-" * 80)
cache_dir = "./cache"
if os.path.exists(cache_dir):
    def get_dir_size(start_path: str) -> int:
        total = 0
        for dirpath, _, filenames in os.walk(start_path):
            for f in filenames:
                fp = os.path.join(dirpath, f)
                try:
                    total += os.path.getsize(fp)
                except OSError:
                    pass
        return total

    before_size = get_dir_size(cache_dir)
    repeated_query = queries_to_test[0]
    times: List[float] = []
    if simple_agent_available:
        for i in range(3):
            r = run_agent(simple_agent, repeated_query)
            times.append(r["elapsed"])
    after_size = get_dir_size(cache_dir)

    if times:
        print(f"Repeated query times: {[f'{t:.2f}' for t in times]}s")
        if len(times) >= 2:
            speedup = times[0] / max(1e-9, statistics.mean(times[1:]))
            print(f"Estimated speedup after first call: {speedup:.2f}x")
    print(f"Cache dir size changed: {before_size} → {after_size} bytes")
else:
    print("No cache directory found; skipping disk cache growth check.")

# 4) Production Readiness Testing
print("\n🏭 Production Readiness Testing")
print("-" * 80)
# 4a) Invalid PDF path
try:
    _ = ProductionRAGChain(
        file_path="./data/THIS_FILE_DOES_NOT_EXIST.pdf",
        chunk_size=1000,
        chunk_overlap=100,
        embedding_model="text-embedding-3-small",
        llm_model="gpt-4.1-mini",
        cache_dir="./cache",
    )
    print("Invalid PDF test: unexpectedly succeeded")
except Exception as e:
    print(f"Invalid PDF test: correctly failed → {str(e)[:160]}…")

# 4b) Missing Tavily API key (tool availability changes)
old_tavily = os.environ.get("TAVILY_API_KEY")
try:
    if old_tavily:
        del os.environ["TAVILY_API_KEY"]
    agent_no_tavily = None
    if rag_chain_available:
        agent_no_tavily = create_langgraph_agent(
            model_name="gpt-4.1-mini",
            temperature=0.1,
            rag_chain=rag_chain,
        )
    if agent_no_tavily:
        # Run a current-events style query to see that Tavily isn't used
        q = "What happened this week in AI policy?"
        r = run_agent(agent_no_tavily, q)
        print(f"No-Tavily agent: {r['elapsed']:.2f}s | tools: {r['tools_used']}")
finally:
    if old_tavily:
        os.environ["TAVILY_API_KEY"] = old_tavily

print("\n✅ Activity #2 complete.")
````

Now we'll integrate **Guardrails AI** into our production system to ensure our agents operate safely and within acceptable boundaries. Guardrails provide essential safety layers for production LLM applications by validating inputs, outputs, and behaviors.

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

**🏢 Enterprise Requirements:**
- **Compliance**: Meet regulatory requirements for data protection
- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses
- **Quality Assurance**: Ensure factual accuracy and relevance

**⚡ Technical Advantages:**
- **Layered Defense**: Multiple validation stages for robust protection
- **Selective Enforcement**: Different guards for different use cases
- **Performance Optimization**: Fast validation without sacrificing accuracy
- **Integration Ready**: Works seamlessly with LangGraph agent workflows

### Demonstrating Core Guardrails

Let's explore the key Guardrails that we'll integrate into our production agent system:

````python
if guardrails_available:
    print("🛡️ Setting up production Guardrails...")
    
    # 1. Topic Restriction Guard - Keep conversations focused on student loans
    topic_guard = Guard().use(
        RestrictToTopic(
            valid_topics=["student loans", "financial aid", "education financing", "loan repayment"],
            invalid_topics=["investment advice", "crypto", "gambling", "politics"],
            disable_classifier=True,
            disable_llm=False,
            on_fail="exception"
        )
    )
    print("✓ Topic restriction guard configured")
    
    # 2. Jailbreak Detection Guard - Prevent adversarial attacks
    jailbreak_guard = Guard().use(DetectJailbreak())
    print("✓ Jailbreak detection guard configured")
    
    # 3. PII Protection Guard - Protect sensitive information
    pii_guard = Guard().use(
        GuardrailsPII(
            entities=["CREDIT_CARD", "SSN", "PHONE_NUMBER", "EMAIL_ADDRESS"], 
            on_fail="fix"
        )
    )
    print("✓ PII protection guard configured")
    
    # 4. Content Moderation Guard - Keep responses professional
    profanity_guard = Guard().use(
        ProfanityFree(threshold=0.8, validation_method="sentence", on_fail="exception")
    )
    print("✓ Content moderation guard configured")
    
    # 5. Factuality Guard - Ensure responses align with context
    factuality_guard = Guard().use(
        LlmRagEvaluator(
            eval_llm_prompt_generator=HallucinationPrompt(prompt_name="hallucination_judge_llm"),
            llm_evaluator_fail_response="hallucinated",
            llm_evaluator_pass_response="factual", 
            llm_callable="gpt-4.1-mini",
            on_fail="exception",
            on="prompt"
        )
    )
    print("✓ Factuality guard configured")
    
    print("\\n🎯 All Guardrails configured for production use!")
    
else:
    print("⚠ Skipping Guardrails setup - not available")
````
